```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option("display.max_columns", None)
sns.set_style("whitegrid")
```

```python
df=pd.read_csv(r"../../dataset/restaurant.csv")
df=df.dropna(subset=["Cuisines"])
df.isnull().sum()
df.head(100)
```

```python
df.select_dtypes(include=["int64","float64"]).columns
```

```python
df.describe()
```

```python
numerical_columns=["Aggregate rating","Average Cost for two","Votes"]
for col in numerical_columns:
    print(f"\n{col}")
    print(f"mean:{df[col].mean():.2f}")
    print(f"median:{df[col].median():.2f}")
    print(f"std:{df[col].std():.2f}")
```

```python
Categorical_columns=["Country Code","City","Cuisines"]
for col in Categorical_columns:
    
    print(f"\n{df[col].value_counts().head(20)}")
```

```python
plt.figure(figsize=(8,5))
sns.countplot(
    data=df,
    x="Country Code",
    order=df["Country Code"].value_counts().index
)
plt.title("Restaurants by Country Code")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Top ten cities with more number of resturants")
top_cities=df["City"].value_counts().head(10)
print(top_cities)
```

```python
plt.figure(figsize=(5,8))
plt.bar(
    x=top_cities.values,
    y=df["City"].value_counts().head(10).index
)
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x=top_cities.values,
    y=top_cities.index
)

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")

plt.show()
```
