```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
df=pd.read_csv(r"../../dataset/restaurant.csv")
df=df.dropna(subset=["Cuisines"])
df.isnull().sum()
```

```python
print(df[["Longitude","Latitude"]].head())
```

```python
print(df[["Longitude","Latitude"]].describe())
```

```python
plt.figure(figsize=(15,8))
plt.scatter(df["Longitude"],
            df["Latitude"],
            c=df["Aggregate rating"],
            cmap="viridis",
            alpha=0.7)
plt.colorbar(label="Aggregate Rating")
plt.show()
```
