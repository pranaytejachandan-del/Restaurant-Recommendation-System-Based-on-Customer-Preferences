```python
def recommend(city, cuisine, price, rating):
    result = df[
        (df["City"] == city) &
        (df["Price range"] <= price) &
        (df["Aggregate rating"] >= rating) &
        (df["Cuisines"].str.contains(cuisine, case=False, na=False))
    ]
    result = result.sort_values(
        by=["Aggregate rating", "Votes"],
        ascending=False
    )
    return result[
        [
            "Restaurant Name",
            "City",
            "Cuisines",
            "Aggregate rating",
            "Votes",
            "Price range"
        ]
    ]
```

```python
recommend(
    city="Noida",
    cuisine="North Indian",
    price=3,
    rating=4.0
)
```
