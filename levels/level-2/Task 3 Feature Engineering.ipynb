```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
df=pd.read_csv(r"../../dataset/restaurant.csv")
df.dropna()
df.isnull().sum()
df
```

```python
CategoricalColumns=df.select_dtypes(include="object").columns
CategoricalColumns
```

```python
l= LabelEncoder()
df["Has Table booking"]=l.fit_transform(df["Has Table booking"])
df["Has Online delivery"]=l.fit_transform(df["Has Online delivery"])
df[["Has Table booking","Has Online delivery"]].head()
```

```python
df[df["Cuisines"]=="North Indian"].head()
```

```python
print(df["Cuisines"].value_counts())
```

```python
df.drop(columns=["Restaurant ID","Restaurant Name","Address","Locality","Locality Verbose"],inplace=True)
```

```python
df.head()
```

```python
df.to_csv("../../dataset/restaurant_cleaned.csv", index=False)
```

```python
O = OneHotEncoder(sparse_output=False)
new = O.fit_transform(df[["Cuisines"]])
new_df=pd.DataFrame(
    new,
    columns=O.get_feature_names_out(["Cuisines"])
)
new_df=df["Cuisines"].str.get_dummies(sep=",")
df=pd.concat([df,new_df],axis=1)
df.drop("Cuisines",axis=1,inplace=True)
```
